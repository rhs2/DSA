# Hash table
## hash table insertion

1. when we insert a key value pair in the hash table
  - 1. first, it will do the hash function: hash(key) to do the hashing
  - 2. we will find the index, index = hash(key) %(remainder) table_size
  - 3. we will check that index slot is empty or not if it is empty we will append that key value in the available index slot


2. if that available slot is already occupied with a data then there is a collision
    - 1. chaining method
    - 2. open addressing method
       - 1. linear probing
       - 2. quadratic probing

1. for collision we can use chaining method, it means in a same bucket we can add up to many element by using linked_list or any other ds

2. for collision in the open addressing we can use these 2 method:

- linear probing method: index = index+1 % table_size and i +=1, first it will look for the right index, if that index is filled up with some value then it will look for next available slot

- quadratic probing method: i =1, index= index+ i*i % table_size and i +=1, but in the quadratic method, it will look for the right index first then it will go into the equation based to look for next available slots 

## hash table deletion
1. apply hash function to find the index
2. check the corresponding slot has the matched key
3. if that matches they then delete the pair
4. if not, then search for if there is a collision exists, if collision exist then handled the key value pair by collision resolution method and mark the slot as deleted
- there can be 3 scenarios
  1. delete key value pair when there is not collision
  2. delete key value pair when there is collision exists by chaining method
  3. delete key value pair when there is collision exists by open addressing method

# hash table search operation
- to retrieve the accosiated value in the hash table
  1. find the index using hash function
  2. check the slot of that correspondes index
  3. if key matches then return the value of that key
  4. if collision occurs then , search through linked_list or use probing method to find the key value pair

## Hash table implementation using list and chaining method


```python
table_size = 10
hash_table = [[] for i in range(table_size)] #to create a hash table we will use list comprehension method

# to find the index we first do hash function and compute it with table_size
def hashing(key):
    return hash(key)%table_size

#insertion in the hash_table
# 1. find the index first
# 1. go through every bucket of the hash table if key matches with its index then update its value
# 2. if corresponding bucket is empty insert the key value pair in the index
def insert(key, value):
    index = hashing(key)
    for item in hash_table[index]:
        if item[0]== key:
            item[1]= value
            return
    hash_table[index].append([key, value])


# serach operation in the hash table
# 1. first we need index of the hash table
# 2. go through every bucket of the hash table
# 3. if given key matches return the value
# 4. else key is not present in the hashtable
def search(key):
    index = hashing(key)
    for item in hash_table[index]:
        if item[0]== key:
            return item[1]
    return None

# deletion operation in hash table
# 1. find the index
# 2. go through every element of the hash table
# 3. if given key== key: delete the value using del keyword
# 4. if key not found return false

def delete(key):
    index = hashing(key)
    for i, item in enumerate(hash_table[index]):
        if item[0]== key:
            del hash_table[index][i]
            return True
    return False

# display the hash table
# 1. loop through hash_table and print it
# 2. here we will also enumerate function to print the index number too

def display():
    for i, item in enumerate(hash_table):
        print("bucket", i, ":", item)

print("before insertion anything in the hash table")
display()
print("after insertion element in the hash table")
insert("sium", 1)
insert("rhs2", 456)
display()
print("value of key sium: ", search('sium'))
print("value of key x: ", search('x'))

if delete("sium"):
    print("deleted")
else:
    print("given key not present in the hash table")

print("after deletion operation")
display()
```

```python
table_size = 10
hash_table = [None] * table_size 
hash_table1= [None for i in range(table_size)] #we can use both 
DELETED = object()

#hash function to find index
def hashing(key):
    return hash(key)%table_size

# insertion operation
# 1. we need key value in parameter
# 2. find the index using hasing function
# 3. if the slot it None or mark as deleted we can insert the key value pair in the slot
# 4. if both conditions are false, if match is found, update the value or if not found use probing method
def insert(key, value):
    index = hashing(key)
    original_index = index #we will use it in future, to perform probing we may lose original index thats why we store that
    while hash_table[index] is not None and hash_table[index] is not DELETED:
        if hash_table[index][0]== key:
            hash_table[index][1]= value
            return
        index = (index+1) % table_size #linear probing
# to prevent the loop here
        if index == original_index:
            raise Exceptions("Hashtable is full")
            # print("Hashtable is full") then return
    
# if both the condition are false then insert the value as tuple
    hash_table[index]= (key, value)

# search operation
# 1. find the index 
# 2. store the original index
# 3. using while loop to see if slots are not None
# 4. check if slots are not deleted and the corresponding index == given key if true then return value
# 5. otherwise use probing
# 6. even if the slots are empty then print given key is not present
def search(key):
    index = hashing(key)
    original_index = index
    while hash_table[index] is not None:
        if hash_table[index] is not DELETED and hash_table[index][0]== key:
            return hash_table[index][1]
        index = (index+1) % table_size # probing
        if index == original_index:
            break    #to prevent the loop
    return None

#1. find the index
# 2. store that original index to a variable
# 3. check if hash table is not None: 
# 4. if true then check if index is deleted or matched with the key
# 5. if matched with given key then mark that index as deleted return true
# 6. another condition is use probing
# 7. to preven loop check index == original index
# 8. outside the loop return False that means given key is not present

def delete(key):
    index = hashing(key)
    original_index= index
    while hash_table[index] is not None:
        if hash_table[index] is not DELETED and hash_table[index][0]== key:
            hash_table[index] = DELETED
            return True
        index = (index+1) % table_size
        if index == original_index:
            break
    return False
    


# display function
# loop through the hash table
# if item is Deleted then print deleted else print the item along with it index
def display():
    for i, item in enumerate(hash_table):
        if item is DELETED:
            print("index",i,": DELETED")
        else:
            print("index", i,": ", item)


print("Before adding any elements")
display()
print('after adding elements in the hash_table')
insert("sium", 1)
insert('rhs2', 456)
insert('rakibul', 13)
display()

print("Value of key sium:", search("sium"))
print("Value of key x:", search("x"))
delete("sium")
display()
```
