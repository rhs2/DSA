# Dijkstra algorithm - invented in 1958 and published in 1959
- here we will talk about dijkstra algo
1. initially we need to store each vertex key and distance value inside that dict(dict comprehension method)
3. we store the vertex key and value as {vertex: inf}
4. we need a graph and edges with distance/cost
5. also a priority queue to insert the first node of the graph then remove it from the queue
6. by the priority queue will pop the elements from the queue
7. it will return 2 value , current_distance and current_node
8. for current distance we need to take all the adj node and weight of that each vertex
9. then find out the new_distance = current_distance + weight
10. if new_distance is < current_distance then update the distance
11. and add that node to queue and we do it untill queue becomes empty
