# to check if graph contains a cycle or not
- in a undirected graph a cycle is ditected if we encounter a node already visited and it is not the immidiate parent node of the current node
## to find the cycle
- 1. to traverse the graph
  2. if a node is actually visited then check that node is the parent node of the current node
  3. if no cycle is detected

```python
# for the  bfs we will use adj list and queue

# UNDIRECTED AND UNWEIGHTED GRAPH
def add_node(v):
    if v in graph:
        print("vertex are already present")
    else:
        graph[v]= []
        
def add_edge(v1, v2):
    if v1 not in graph:
        print("not present in the graph")
    elif v2 not in graph:
        print("not present in the graph")
    else:
        graph[v1].append(v2)
        graph[v2].append(v1)
def del_node(v):
    if v not in graph:
        print("not present in the graph")
    else:
        graph.pop(v)
        for i in graph:
            list1= graph[i]
            if v in list1:
                list1.remove(v)
def del_edge(v1,v2):
    if v1 not in graph:
        print("not present in the graph")
    elif v2 not in graph:
        print("not present in the graph")
    else:
        graph[v1].remove(v2)
        graph[v2].remove(v1)


# for the dfs iterative we will use adj list
# also use stack and visited set to count the nodes
# 1. append that staring node to stack
# 2. pop from stack
# 3. check if popped element already in the visited
# 4. if not print that node and add it to visited set
# 5. loop through graph [current node]
# 6. add those iterative in the stack
def dfs_iterative(node, graph):
    if node not in graph:
        print("not present in the graph")
        return
    stack = []
    stack.append(node)
    visited = set()
    while stack:
        current = stack.pop()
        if current not in visited: 
            print(current, end=" ")
            visited.add(current)
            for i in graph[current]:
                stack.append(i)
# 1. choose a starting node and add them to q
# 2. mark that as visited
#  we need count the parent node thats why we initialize a dict name parent ={node:None}- initial node will be none
# 3. remove node from q, add all the unvisited node of that removed node to q
# 4. check if iterate node is already present in the visited
# 5. if not then add the iterate node to q and mark them as visited also update the parent dict with its loops node(i) and parent node current
# 6. elif if parent node of current != i: return true and outside the loop return false
def bfs(node, graph):
    visited = set()
    if node not in graph:
        print("given not present in the graph")
        return
    q =[]
    parent ={node: None}
    q.append(node)
    visited.add(node)
    while q:
        current = q.pop(0)
        for i in graph[current]:
            if i not in visited:
                parent[i]= current
                q.append(i)
                visited.add(i)
            elif parent[current] != i:
                return True
    return False

graph = {}
add_node("A")
add_node("B")
add_node("C")
add_node("D")
add_node("E")
add_node("F")
add_node("G")
add_node("H")
add_edge("A", "B")
add_edge("A", "C")
add_edge("B", "D")
add_edge("D", "E")
add_edge("E", "F")
add_edge("D", "F")
add_edge("F", "G")
add_edge("G", "H")
add_edge("F", "H")

# if bfs is true then cycle is detected otherwise no 
if bfs("A", graph):
    print("Cycle is detected")
else:
    print("cycle is not detected")
```
